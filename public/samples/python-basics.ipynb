## 1. Printing
`print()` shows things on the screen.

```python
print("Hello, world!")
print("Python is", 30 + 4, "years old")
```

## 2. Variables
A variable is a name for a value. Cells share variables.

```python
favorite_food = "pizza"
slices = 8
print("I could eat", slices, "slices of", favorite_food)
```

```python
# This cell uses the variables from the cell above:
slices_per_friend = slices / 4
slices_per_friend
```

## 3. Asking questions
`input()` waits for you to type an answer.

```python
animal = input("What's your favorite animal? ")
print(f"A {animal}? Great choice!")
```

## 4. Loops
Loops repeat things so you don't have to.

```python
for number in range(1, 6):
    print(number, "squared is", number ** 2)
```
