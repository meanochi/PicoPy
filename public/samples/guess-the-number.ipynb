# Guess the number 🎲

A tiny game: the computer picks a secret number from 1 to 20,
and you try to guess it. Run the cell below to play!

```python
import random

secret = random.randint(1, 20)
tries = 0
print("I'm thinking of a number from 1 to 20...")

while True:
    guess = int(input("Your guess: "))
    tries += 1
    if guess < secret:
        print("Too low! ⬇️")
    elif guess > secret:
        print("Too high! ⬆️")
    else:
        print(f"🎉 Got it in {tries} tries!")
        break
```
